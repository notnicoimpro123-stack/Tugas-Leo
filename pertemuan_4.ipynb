Sampling Distribution

````python
sample_means =[]
````

````python
for i in range(1000):

  sample= np.random.choice(
      nilai,
      size=2,
      replace= False
  )

  mean = np.mean(sample)

  sample_means.append(mean)
````

````python
populasi = np.random.normal(
    loc= 70, #mean
    scale= 5, #std
    size=10000 #banyaknya data
)
````

````python
plt.hist(populasi, bins=30)
plt.title("Distribusi populasi")
plt.xlabel("nilai")
plt.ylabel("frekuensi")
plt.show
````

Scipy

````python
from scipy import stats
````

````python
distribusi_normal = stats.norm.rvs(
    loc=70,
    scale=5,
    size=10000,

)

plt.hist(distribusi_normal, bins=30)
plt.title("Distribusi normal")
plt.xlabel("nilai")
plt.ylabel("frekuensi")
plt.show
````

````python
distribusi_exponential = stats.expon.rvs(
    scale=5,
    size=10000,

)

plt.hist(distribusi_exponential, bins=30)
plt.title("Distribusi normal")
plt.xlabel("nilai")
plt.ylabel("frekuensi")
plt.show
````

````python
distribusi_poisson = stats.poisson.rvs(
    mu=5, #rata rata
    size=10000, #banyaknya data

)

plt.hist(distribusi_poisson, bins=30)
plt.title("Distribusi poisson")
plt.xlabel("nilai")
plt.ylabel("frekuensi")
plt.show
````
