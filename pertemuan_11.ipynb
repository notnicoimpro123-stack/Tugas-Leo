```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
```

```python
iris = load_iris(as_frame=True)
keep = iris.target.isin([1, 2])
x = iris.data.loc[keep,
    ['petal length (cm)', 'petal width (cm)']]
y = (iris.target.loc[keep] == 2).astype(int)
print(x.shape, y.value_counts().to_dict())
```

```python
x_tr, x_te, y_tr, y_te = train_test_split(
    x, y, test_size=0.20,
    random_state=42, stratify=y
)
print(len(y_tr), len(y_te))
```

```python
nb = GaussianNB()
nb.fit(x_tr, y_tr)
yhat_nb = nb.predict(x_te)
p_nb = nb.predict_proba(x_te)[:,1]
print(yhat_nb[:5], p_nb[:5])
```

```python
lr = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=1000,
                       random_state=42)
)
lr.fit(x_tr, y_tr)
yhat_lr = lr.predict(x_te)
p_lr = lr.predict_proba(x_te)[:,1]
```

```python
from sklearn.metrics import (accuracy_score,
    precision_score, recall_score, f1_score)
for name,pred in [('NB',yhat_nb),('LR',yhat_lr)]:
  print(name, accuracy_score(y_te,pred),
        precision_score(y_te,pred),
        recall_score(y_te,pred),
        f1_score(y_te,pred))
```
