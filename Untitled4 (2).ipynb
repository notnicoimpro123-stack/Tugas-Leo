```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

cell 2:data sintesis

```python
np.random.seed(2008)
n_samples = 1000

x1 = np.random.rand(n_samples) * 2
x2 = np.random.rand(n_samples) * 3

noise = np.random.rand(n_samples) * 1.5

#pengaruh linear : Y = 5 + 3,5*X1 - 2 *X2 + noise
Y = 5 + 3.5 * x1 - 2 * x2 + noise

#menggabungkan menjadi data frame
df = pd.DataFrame({'X1': x1, 'X2': x2, 'Y': Y})
display(df.head())
```

cell 3:mode fit

```python
x = df[['X1', 'X2']]
y = df['Y']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=2008)

model = LinearRegression()
model.fit(x_train, y_train)

print(f"Intercept (konstanta): {model.intercept_:.4f}")
print(f"Koefisien x1: {model.coef_[0]:.4f}")
print(f"Koefesien x2: {model.coef_[1]:.4f}")

# Y = 5,71 + 3,47 X1 - 1,98 X2
```

cell 4: prediksi dan matriks

```python
y_pred = model.predict(x_test)

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Mean Squared Error: {mse:.4f}")
print(f"R-squared (R2): {r2:.4f}")
```

cell 5: visualisasi

```python
plt.figure(figsize=(8,6))

sns.scatterplot(x=y_test, y=y_pred, color = 'blue', alpha =  0.7)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], color = 'red', linestyle='--', lw=2 )
plt.title('Nilai aktual vs prediksi')
plt.xlabel('Nilai aktual (y)')
plt.ylabel('Nilai prediksi (y_topi)')
plt.show()
```
