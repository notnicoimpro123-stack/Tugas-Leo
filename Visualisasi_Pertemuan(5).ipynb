```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

distribusi_normal = stats.norm.rvs(
    loc=70, #mean
    scale=5, #std deviasi
    size=10000 #jumlah sampel
)

plt.hist(distribusi_normal,bins=30)
plt.title("Distribusi Normal")
plt.xlabel("Nilai")
plt.ylabel("Frekuensi")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

distribusi_normal = stats.norm.rvs(
    loc=70, #mean
    scale=5, #std deviasi
    size=10000 #jumlah sampel
)

plt.hist(distribusi_normal, bins=30, density=True, cumulative=True,)
plt.title("(CDF) - Distribusi Normal")
plt.xlabel("Nilai")
plt.ylabel("Cumulative Probability")
plt.show()
```

```python
from scipy import stats

distribusi_longtail = stats.expon.rvs(
    scale=5, #std deviasi
    size=10000 #jumlah sampel
)

plt.hist(distribusi_longtail, bins=30)
plt.title("Distribusi Long-Tailed")
plt.xlabel("Nilai")
plt.ylabel("Frekuensi")
plt.show()
```
