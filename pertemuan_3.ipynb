```python
# kita akan membuat populasi nilai 1000 mahasiswa
# dengan metode distribusi normal yang random
# dengan rata-rata adalah 70
# dengan standar deviasi adalah 10
# dengan kode random 123

np.random.seed(123)

populasi = np.random.normal(
    loc = 70,
    scale = 10,
    size = 1000
)

populasi
```

---

```python
# Random sampling

sampel = np.random.choice(
    populasi,
    size = 30,
    replace = False

)

sampel
```

```python
np.mean(populasi)
```

```python
np.mean(sampel)
```

---

```python
sampel_acak = np.random.choice(
    populasi,
    size = 50,
    replace = False

)

sampel_acak
```

```python
#Bias

sampel_bias = populasi[populasi > 80][:50]

print("populasi : ", np.mean(populasi))
print("random :", np.mean(sampel_acak))
print("bias :", np.mean(sampel_bias))
```
